```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import rioxarray as rxr

```

```python
PCT_CHANGE_DNBR_RBR_URL = "https://burn-severity-backend-prod.s3.us-east-2.amazonaws.com/public/DSE/Dove/pct_change_dnbr_rbr.tif"

pct_change_dnbr_rbr = rxr.open_rasterio(PCT_CHANGE_DNBR_RBR_URL, masked=True).squeeze()
```

```python
fig, ax = plt.subplots(figsize=(13, 5))

pct_change_dnbr_rbr.plot(
    ax=ax,
    cmap="RdYlBu_r",
    vmin=-20,
    vmax=20,
    cbar_kwargs={"label": "(RBR - DNBR) / DNBR"},
)

ax.set_title("Dove Fire: Pct Change in RBR vs DNBR")
ax.set_xlim([-118.16, -118.09])
ax.set_ylim([35.445, 35.465])
plt.show()
```
